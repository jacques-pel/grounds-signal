## Make the API Call

### Define function to perform Text Search Call

```python
# Google Places Text Search API Call

def text_search(query, page_token=None):
    url = "https://places.googleapis.com/v1/places:searchText"


    # Main guidelines for the call
    headers = {
        "Content-Type": "application/json",
        "X-Goog-Api-Key": api_key,
        "X-Goog-FieldMask": "places.id,places.displayName,places.formattedAddress,places.location,places.rating,places.userRatingCount,places.priceLevel,places.reviews,places.primaryType,places.outdoorSeating,places.servesCoffee,places.goodForGroups,places.liveMusic,places.reservable,places.photos,places.priceRange,places.types,places.takeout,nextPageToken,places.editorialSummary,places.generativeSummary,places.goodForChildren,places.neighborhoodSummary,places.reviewSummary"
    }


    # Details for the call: query, number of results per page, region, possibly type
    body = {
        "textQuery": query,
        "pageSize": 20,
        "regionCode": "JP",
        #"includedType": "coffee_shop"   May include later
    }


    # Used to track if page we are pulling from each call: 20 items per call, 3 pages max
    if page_token:
        body["pageToken"] = page_token


    # Store the response
    response = requests.post(url, headers=headers, json=body)
    return response.json()
```

### Define function to iterate through Text Search page tokens

```python
def collect_places(query):
    all_places = []
    next_page_token = None

    for page in range(3):
        print(f"Fetching page {page + 1}...")
        result = text_search(query, page_token=next_page_token)

        places = result.get("places", [])
        all_places.extend(places)
        print(f"Got {len(places)} results. Total so far: {len(all_places)}")

        next_page_token = result.get("nextPageToken")
        if not next_page_token:
            print("No more pages.")
            break

    return all_places
```

```python
harajuku_places = collect_places("coffee shop in Harajuku, Tokyo")
```
